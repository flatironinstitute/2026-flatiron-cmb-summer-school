## Introduction: B-modes and Primordial Gravitational Waves

From the [CMB Polarization notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_CMB_Polarization.ipynb), we have learnt that the polarization field of the CMB can be uniquely decomposed into two components of opposite parity: **E-modes** (gradient-like, curl-free patterns) and **B-modes** (curl-like, divergence-free patterns). This decomposition is powerful because of a simple physical fact: **scalar (density) perturbations — the ones that seed all cosmic structure — produce no B-modes at linear order**. Any primordial degree-scale $B$-mode is therefore a window onto other physics, beyond the scalar density perturbations that dominate the temperature and $E$-mode skies. 

The most exciting potential source of primordial B-modes is a background of **primordial gravitational waves (PGWs)**: tensor perturbations of the metric that essentially all models of **cosmic inflation** predict, generated by quantum fluctuations of spacetime itself during a rapid exponential expansion at the beginning of the Universe. As these gravitational waves stream through the plasma at recombination, they imprint a characteristic B-mode polarization pattern peaking at degree angular scales.

The amplitude of this signal is parameterized by **the tensor-to-scalar ratio $r$**: the ratio of the primordial power in tensor perturbations to that in scalar perturbations at a pivot scale, conventionally $k = 0.05,\mathrm{Mpc}^{-1}$. This number is powerful because it turns a *pattern on the sky into a measurement of the physics of the very early Universe*. In simple inflationary models, $r$ is directly tied to the expansion rate and energy scale during inflation,
$$
V^{1/4} \approx 1.04 \times 10^{16},\mathrm{GeV},
\left(\frac{r}{0.01}\right)^{1/4},
$$
A **detection of primordial $B$-modes** ($r > 0$) would be evidence that inflation happened, but it would also mean that we had seen the faint imprint of gravitational waves produced when the Universe was almost unimaginably young. In that sense, the CMB would be acting like a fossil record of spacetime itself. In the inflationary picture, those waves began as quantum fluctuations of spacetime itself, later stretched across the sky by cosmic expansion. The size of the signal, described by $r$, would tell us how energetic and rapid that early expansion was — probing physics at energies far beyond anything we can recreate in a laboratory, and providing one of the few observational windows onto the quantum origin of spacetime fluctuations. It would also help distinguish between different possible versions of inflation, showing which kinds of early-Universe physics are still viable and which are ruled out. 


![](https://images.newscientist.com/wp-content/uploads/2019/02/mg32180201.jpg)

### Why is this measurement so hard? 

Current data already require this signal to be very small. The most stringent upper limits, from the [BICEP/Keck + Planck + WMAP](https://arxiv.org/pdf/2110.00483) gives $r_{0.05}<0.035$ at 95% confidence. Upcoming experiments, such as the Simons Observatory and LiteBIRD aim to push sensitivity to the $\sigma(r)\sim 0.001-0.003$ level, where either a detection would open a direct observational window onto ultra-high-energy physics, or a non-detection would rule out large families of otherwise plausible early-Universe models.

If the signal exists at the currently allowed level, it is buried under everything else:

1. **CMB temperature and E-mode polarization** – scalar density perturbations produce the dominant temperature anisotropies and E-mode polarization, but no B-modes at linear order. This is what makes B-modes such a clean target in principle — but in practice it also means that any leakage from T or E into B, due to masking, filtering, beam effects or calibration errors, has to be controlled extremely carefully.
2. **Gravitational lensing** of E-modes by large-scale structure converts a small fraction of E into B, producing a guaranteed "lensing B-mode" signal that dominates over the primordial one at $\ell \gtrsim 100$ (see the [Lensing Simulation](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Lensing_Simulation.ipynb) and [Lensing Reconstruction](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Lensing_Reconstruction.ipynb) notebooks).
3. **Galactic foregrounds** — polarized thermal dust and synchrotron emission from our own Galaxy — are brighter than the primordial signal at *every* frequency and over most of the sky.
4. **Instrumental noise and systematics** must be controlled at the nanokelvin level (see the [Modeling Instrumental Noise and Beams](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_modeling_noise_and_beams.ipynb) and [Real Experiment Beams](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_Beam_Systematics.ipynb) notebooks).

## Section 1: The CMB B-mode Power Spectrum

At the level allowed by the data, any primordial B-mode signal generated by PGWs, parameterized by the tensor-to-scalar ratio $r$, is mostly dominated by a lensing B-mode signal, due to gravitational lensing of E-modes into B-modes, parameterized by the lensing amplitude $A_\mathrm{lens}$ (=1 in standard $\Lambda$CDM) (see the [Lensing Simulation notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Lensing_Simulation.ipynb) for where this signal comes from).

The total CMB B-mode angular power spectrum is:
$$D_\ell^{BB} = r \cdot D_\ell^{BB,\mathrm{tens}} + A_\mathrm{lens} \cdot D_\ell^{BB,\mathrm{lens}}$$

where $D_\ell = \ell(\ell+1)C_\ell / (2\pi)$ [µK²], and:
- $D_\ell^{BB,\mathrm{tens}}$ is the primordial tensor contribution for $r=1$ (computed by CAMB)
- $D_\ell^{BB,\mathrm{lens}}$ is the lensing B-mode spectrum (computed by CAMB)

```python
# ============================================================
# 1.1  Compute CMB power spectra with CAMB
# ============================================================

lmax_camb = 1000

# Standard LCDM cosmology
cosmo_params = dict(H0=67.5, ombh2=0.022, omch2=0.122, mnu=0.06, omk=0, tau=0.06)

# --- Lensed scalar BB (r=0, gives lensing B-modes) ---
# This part is the same as what is done in CMB_School_Simulating_the_CMB.ipynb
pars0 = camb.CAMBparams()
pars0.set_cosmology(**cosmo_params)
pars0.WantTensors = False
pars0.InitPower.set_params(As=2e-9, ns=0.965, r=0)
pars0.set_for_lmax(lmax_camb, lens_potential_accuracy=1)
results0 = camb.get_results(pars0)
cls_lens = results0.get_lensed_scalar_cls(lmax=lmax_camb, CMB_unit='muK')
# Shape: (lmax_camb+1, 4), columns TT, EE, BB, TE; index = ell

# --- Tensor BB for r=1 ---
# We add the tensor component
pars1 = camb.CAMBparams()
pars1.set_cosmology(**cosmo_params)
pars1.WantTensors = True
pars1.InitPower.set_params(As=2e-9, ns=0.965, r=1, nt=0)
pars1.set_for_lmax(lmax_camb, lens_potential_accuracy=0)
results1 = camb.get_results(pars1)
cls_tens = results1.get_tensor_cls(lmax=lmax_camb, CMB_unit='muK')
# Tensor D_ell for r=1

# Store D_ell arrays indexed by ell (row index = ell)
ell_camb   = np.arange(lmax_camb + 1)   # 0..lmax_camb
Dl_lens_BB = cls_lens[:, 2]             # D_ell^BB from lensing
Dl_tens_BB = cls_tens[:, 2]             # D_ell^BB from tensors (r=1)

# Interpolation functions for arbitrary ell
f_Dl_lens = interp1d(ell_camb, Dl_lens_BB, bounds_error=False, fill_value=0.)
f_Dl_tens = interp1d(ell_camb, Dl_tens_BB, bounds_error=False, fill_value=0.)

def cmb_Dl_BB(ell, r=0.01, A_lens=1.0):
    """Total CMB D_ell^BB [muK^2] for given r and lensing amplitude."""
    return r * f_Dl_tens(ell) + A_lens * f_Dl_lens(ell)

def cmb_Cl_BB(ell, r=0.01, A_lens=1.0):
    """Total CMB C_ell^BB [muK^2] = D_ell * 2*pi / (ell*(ell+1))."""
    Dl = cmb_Dl_BB(ell, r, A_lens)
    return Dl * 2 * np.pi / (ell * (ell + 1))

print(f'CAMB spectra computed up to ell={lmax_camb}')
print(f'Peak lensing D_ell at ell ~ {ell_camb[np.argmax(Dl_lens_BB)]}')
```

```python
# ============================================================
# 1.2  Plot the CMB B-mode power spectrum
# ============================================================

ell_plot = np.arange(2, 400)

fig, ax = plt.subplots(figsize=(7, 5))

ax.plot(ell_plot, f_Dl_lens(ell_plot), 'k--', lw=2.5, label=r'Lensing BB ($A_{\rm lens}=1$)')
for r_val, color in [(0.1, 'C0'), (0.01, 'C1'), (0.001, 'C3')]:
    ax.plot(ell_plot,
            r_val * f_Dl_tens(ell_plot),
            color=color, label=f'Primordial tensors, $r={r_val}$')

ax.set_yscale('log')
ax.set_xscale('log')
ax.set_xlabel(r'Multipole $\ell$')
ax.set_ylabel(r'$D_\ell^{BB}$ [$\mu{\rm K}^2$]')
ax.set_title('CMB B-mode Power Spectrum')
ax.legend(loc='upper left')
ax.set_xlim(2, 400)
ax.set_ylim(1e-6, 0.2)
ax.axvspan(30, 300, alpha=0.07, color='grey', label='Ground-based analysis range')
plt.tight_layout()
plt.show()
```

The primordial $B$-mode signal has a distinctive shape: it is expected to **peak on degree angular scales** – the "recombination bump" at $\ell \approx 80$, with another peak on the largest scales, $\ell \lesssim 10$, from reionization. But we do not know how large the signal should be. The parameter $r$ is currently constrained only by upper limits, and different inflationary models predict different amplitudes depending on different possible mechanisms for the earliest expansion of the Universe.

The primordial signal is below the lensing $B$-mode signal over much of the range $\ell \gtrsim 80$ for $r=0.01$. Lensing $B$-modes are therefore a kind of cosmological foreground: they are interesting physics in their own right, but they sit in front of the primordial signal we are trying to measure.

The challenge is that both the primordial signal and lensing signal are **subdominant compared to galactic foreground emission** at most frequencies and angular scales. We need to measure the sky at multiple frequencies to separate them.

## Section 2: Galactic Foregrounds

Two galactic emission processes dominate the B-mode signal over most of the sky and frequency range. The frequency dependence of foreground emission is given by its Spectral Energy Distribution (SED), which tells us how the brightness or polarised intensity of a non-CMB component changes with observing frequency.

### 2a. Thermal Dust Emission
Magnetically aligned dust grains emit thermally polarized radiation, well described by a **modified blackbody** (MBB) spectral energy distribution:
$$f_d(\nu) \propto \nu^{\beta_d + 1} \cdot \frac{B_\nu(T_d)}{B_{\nu_0}(T_d)},$$
where $B_\nu(T)$ is the Planck function, $\beta_d \approx 1.54$, $T_d \approx 20$ K, and $\nu_0 = 353$ GHz (the Planck reference frequency). This SED rises steeply with frequency, making dust dominant above ~100 GHz.

### 2b. Synchrotron Emission
Relativistic electrons spiraling in the galactic magnetic field emit a **power-law** SED in Rayleigh-Jeans temperature:
$$f_s(\nu) \propto \left(\frac{\nu}{\nu_0}\right)^{\beta_s},$$
with $\beta_s \approx -3.0$ and $\nu_0 = 23$ GHz. Synchrotron dominates at low frequencies (<70 GHz).

#### Unit Conventions
We work throughout in $K_\mathrm{CMB}$ units: the signal is expressed as a fluctuation in CMB thermodynamic temperature. The conversion from Rayleigh-Jeans ($K_\mathrm{RJ}$) to $K_\mathrm{CMB}$ introduces a frequency-dependent factor:
$$g(\nu) = \frac{dT_\mathrm{CMB}}{dT_\mathrm{RJ}} = \left(\frac{h\nu}{k_\mathrm{B} T_\mathrm{CMB}}\right)^2 \frac{e^x}{(e^x - 1)^2}, \quad x = \frac{h\nu}{k_\mathrm{B} T_\mathrm{CMB}}.$$
The **CMB SED in $K_\mathrm{CMB}$ units is frequency-independent ($=1$), which is the key property we exploit for component separation**.

```python
# ============================================================
# 2.1  Foreground SED functions (in K_CMB units)
# ============================================================

def g_nu(nu_GHz):
    """K_RJ -> K_CMB unit conversion factor: g(nu) = (x^2 e^x)/(e^x-1)^2."""
    x  = h_P * nu_GHz * 1e9 / (k_B * T_CMB)
    ex = np.exp(x)
    return (x**2 * ex) / (ex - 1)**2

def dust_sed(nu_GHz, nu0_GHz=353., beta_d=1.54, T_d=20.):
    """
    Thermal dust SED in K_CMB, normalized to 1 at nu0.
    Modified blackbody: f(nu) ~ nu^(beta+1) * B_nu(T_d)  in K_RJ.
    Then converted to K_CMB via g(nu0)/g(nu).
    """
    x  = h_P * nu_GHz * 1e9 / (k_B * T_d)
    x0 = h_P * nu0_GHz * 1e9 / (k_B * T_d)
    sed_rj = (nu_GHz / nu0_GHz)**(beta_d + 1) * np.expm1(x0) / np.expm1(x)
    return sed_rj * g_nu(nu0_GHz) / g_nu(nu_GHz)

def synch_sed(nu_GHz, nu0_GHz=23., beta_s=-3.0):
    """
    Synchrotron SED in K_CMB, normalized to 1 at nu0.
    Power law in K_RJ: f(nu) ~ (nu/nu0)^beta_s.
    Then converted to K_CMB via g(nu0)/g(nu).
    """
    sed_rj = (nu_GHz / nu0_GHz)**beta_s
    return sed_rj * g_nu(nu0_GHz) / g_nu(nu_GHz)

# --- Verify: at reference frequencies, SEDs = 1 ---
print(f'dust_sed(353 GHz)  = {dust_sed(353.):.4f}  (expect 1.0)')
print(f'synch_sed(23 GHz)  = {synch_sed(23.):.4f}   (expect 1.0)')
print(f'CMB SED at all nu  = 1.0  (frequency-independent in K_CMB)')
```

```python
# ============================================================
# 2.2  Plot SEDs in K_CMB units
# ============================================================

nu_arr = np.linspace(20, 400, 300)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Left: SEDs
ax = axes[0]
ax.semilogy(nu_arr, np.abs(dust_sed(nu_arr)), 'C1', label=r'Thermal dust ($\beta_d=1.54,\, T_d=20$ K)')
ax.semilogy(nu_arr, np.abs(synch_sed(nu_arr)), 'C0', label=r'Synchrotron ($\beta_s=-3.0$)')
ax.axhline(1.0, color='k', ls='--', label='CMB (flat in $K_{\\rm CMB}$)')

# Mark our frequency bands
band_freqs = [93, 145, 217, 353]
for nu in band_freqs:
    ax.axvline(nu, color='grey', ls=':', alpha=0.6)

ax.set_xlabel(r'Frequency [GHz]')
ax.set_ylabel(r'SED [$K_{\rm CMB}$ / $K_{\rm CMB}$ at ref. freq.]')
ax.set_title('Spectral Energy Distributions')
ax.legend(fontsize=10, loc='center right')
ax.set_xlim(20, 400)

# Right: SED sensitivity to spectral index
ax2 = axes[1]
for beta_d_val, ls in [(1.3, '--'), (1.54, '-'), (1.8, ':')]:
    ax2.semilogy(nu_arr, np.abs(dust_sed(nu_arr, beta_d=beta_d_val)),
                 'C1', ls=ls, label=f'$\\beta_d={beta_d_val}$')
for beta_s_val, ls in [(-2.7, '--'), (-3.0, '-'), (-3.3, ':')]:
    ax2.semilogy(nu_arr, np.abs(synch_sed(nu_arr, beta_s=beta_s_val)),
                 'C0', ls=ls, label=f'$\\beta_s={beta_s_val}$')
for nu in band_freqs:
    ax2.axvline(nu, color='grey', ls=':', alpha=0.6)
ax2.set_xlabel(r'Frequency [GHz]')
ax2.set_ylabel(r'SED [$K_{\rm CMB}$]')
ax2.set_title('SED Dependence on Spectral Index')
ax2.legend(fontsize=9)
ax2.set_xlim(20, 400)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 2.3  Foreground angular power spectra (B-mode)
# ============================================================

# Power-law D_ell models: D_ell = A * (ell/ell0)^alpha
ell0 = 80.   # pivot scale

def dust_Dl_BB(ell, A_d=20., alpha_d=-0.4):
    """Dust BB power spectrum D_ell [muK^2] at nu0=353 GHz."""
    return A_d * (ell / ell0)**alpha_d

def synch_Dl_BB(ell, A_s=2., alpha_s=-0.8):
    """Synchrotron BB power spectrum D_ell [muK^2] at nu0=23 GHz."""
    return A_s * (ell / ell0)**alpha_s

def Dl_to_Cl(Dl, ell):
    """D_ell [muK^2] -> C_ell [muK^2]: divide by ell(ell+1)/(2 pi)."""
    return Dl * 2 * np.pi / (ell * (ell + 1))

def Cl_to_Dl(Cl, ell):
    """C_ell [muK^2] -> D_ell [muK^2]."""
    return Cl * ell * (ell + 1) / (2 * np.pi)

# --- Plot foreground spectra vs CMB ---
ell_pl = np.arange(20, 400)

# Dust at 150 GHz (scale from 353 GHz reference)
f_d_150 = dust_sed(150.)
Dl_dust_150 = dust_Dl_BB(ell_pl) * f_d_150**2

# Synch at 90 GHz
f_s_90 = synch_sed(90.)
Dl_synch_90 = synch_Dl_BB(ell_pl) * f_s_90**2

fig, ax = plt.subplots(figsize=(7, 5))
ax.loglog(ell_pl, cmb_Dl_BB(ell_pl, r=0.01), 'k-',  lw=2.5, label=r'CMB: lensing + $r=0.01$')
ax.loglog(ell_pl, f_Dl_lens(ell_pl),           'k--', lw=1.5, label=r'CMB: lensing only')
ax.loglog(ell_pl, Dl_dust_150,                  'C1',          label='Dust at 150 GHz')
ax.loglog(ell_pl, Dl_synch_90,                  'C0',          label='Synchrotron at 90 GHz')
ax.set_xlabel(r'Multipole $\ell$')
ax.set_ylabel(r'$D_\ell^{BB}$ [$\mu{\rm K}^2$]')
ax.set_title('B-mode Power Spectra: CMB vs Foregrounds')
ax.legend()
ax.set_xlim(20, 400)
ax.set_ylim(1e-5, 5)
plt.tight_layout()
plt.show()
```

**Key insight**: At 150 GHz and ell~80, dust is ~100x larger than the r=0.01 primordial signal!

## Section 3: The Multi-Frequency Signal Model

### 3.1 The Frequency–Frequency Cross-Power Spectrum

When we observe the sky at frequencies $\nu_i$ and $\nu_j$, the observed B-mode cross-power spectrum is:
$$C_\ell^{B\nu_i \times B\nu_j} = C_\ell^{\rm CMB} + f_d(\nu_i)\, f_d(\nu_j)\, C_\ell^{\rm dust} + f_s(\nu_i)\, f_s(\nu_j)\, C_\ell^{\rm synch} + N_\ell^{\nu_i \nu_j},$$

where:
- $C_\ell^{\rm CMB} = r C_\ell^{\rm tens} + A_{\rm lens} C_\ell^{\rm lens}$ is frequency-independent (in $K_{\rm CMB}$)
- $f_d(\nu_i)$, $f_s(\nu_i)$ are the foreground SEDs at band $i$
- $C_\ell^{\rm dust}$, $C_\ell^{\rm synch}$ are the component power spectra
- $N_\ell^{\nu_i \nu_j} = \delta_{ij} N_\ell^{\nu_i}$ is the noise (diagonal, uncorrelated between bands)

### 3.2 The SED Mixing Matrix

The outer product $f(\nu_i) \cdot f(\nu_j)$ defines the **SED mixing matrix** for each component. For $N_\nu$ frequency bands, this is an $N_\nu \times N_\nu$ matrix that tells us how each component scales across frequency–frequency pairs. Its structure is key to component separation:
- CMB: flat mixing matrix (all entries = 1)
- Dust: matrix entries increase steeply toward high frequencies
- Synchrotron: matrix entries increase toward low frequencies

By fitting a parametric model to the multi-frequency data vector $\{C_\ell^{\nu_i \times \nu_j}\}$, we can simultaneously constrain $r$ and the foreground parameters.

This *parametric, power-spectrum-level* approach is complementary to the *blind, map-level* Internal Linear Combination (ILC) technique used in the [Y-map notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_ymap.ipynb): the ILC combines maps with weights that minimize the variance of the residual, without assuming a foreground model, while here we fit an explicit physical model for every component directly at the power-spectrum level.


```python
# ============================================================
# 3.1  Define frequency bands and compute SED mixing matrices
# ============================================================

freqs      = np.array([93., 145., 217., 353.])   # GHz
band_names = ['93 GHz', '145 GHz', '217 GHz', '353 GHz']
n_freq     = len(freqs)

# Fiducial spectral indices
beta_d_fid = 1.54
beta_s_fid = -3.0

# SED vectors at each band
f_dust  = dust_sed(freqs, beta_d=beta_d_fid)   # shape (n_freq,)
f_synch = synch_sed(freqs, beta_s=beta_s_fid)  # shape (n_freq,)
f_cmb   = np.ones(n_freq)                       # CMB: frequency-independent

print('SED values at each band (K_CMB / K_CMB_ref):')
print(f'{"Band":>10}  {"CMB":>8}  {"Dust":>10}  {"Synch":>12}')
for i, nu in enumerate(freqs):
    print(f'{nu:>8.0f} GHz  {f_cmb[i]:>8.4f}  {f_dust[i]:>10.4f}  {f_synch[i]:>12.6f}')
```

## Section 4: Synthetic Data Generation

We now build the complete signal model and generate a realistic synthetic dataset.

### 4.1 Bandpowers and covariance

In practice, we work with **bandpower** averages over multipole bins $b = [\ell_\min^b, \ell_\max^b]$:
$$\hat{C}_b^{\nu_i \times \nu_j} = \sum_\ell w_b(\ell)\, C_\ell^{\nu_i \times \nu_j},$$
with weights $w_b(\ell) \propto (2\ell+1)$ for uniform-weight binning.

The **bandpower covariance** for independent ell bins is:
$$\mathrm{Cov}\left[\hat{C}_b^{ij},\, \hat{C}_b^{kl}\right] = \frac{C_b^{ik,\mathrm{tot}} C_b^{jl,\mathrm{tot}} + C_b^{il,\mathrm{tot}} C_b^{jk,\mathrm{tot}}}{N_{\rm modes}^b},$$
where $N_{\rm modes}^b = (2\ell_b + 1)\, f_{\rm sky}\, \Delta\ell_b$ counts the number of independent modes per bin, $f_{\rm sky}$ is the sky fraction observed, and $C_b^{\cdot,\mathrm{tot}}$ includes both signal and noise.

For cross-spectra between different frequency bands ($i \neq j$), noise contributes only through the total signal+noise in the variance terms (not as a noise bias in the mean), making cross-spectra preferred in practice.

In a real analysis the bandpowers are estimated from masked, noisy maps, which introduces mode coupling, noise bias and transfer functions — the machinery covered in the [Power Spectrum Analysis notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Power_Spectrum_Analysis.ipynb). Here we skip the map-making stage and simulate the bandpowers directly, using the noise + beam model introduced in the [Modeling Instrumental Noise and Beams notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_modeling_noise_and_beams.ipynb).


```python
# ============================================================
# 4.1  Signal model: C_ell(nu_i, nu_j) for all frequency pairs
# ============================================================

# Multipole binning
ell_min, ell_max = 30, 300
n_ell_bins = 9
ell_edges  = np.linspace(ell_min, ell_max, n_ell_bins + 1)
ell_bins   = 0.5 * (ell_edges[:-1] + ell_edges[1:])   # bin centers
delta_ell  = ell_edges[1] - ell_edges[0]               # bin width = 30

# Upper-triangle indices (unique freq-freq pairs)
i_triu, j_triu = np.triu_indices(n_freq)   # all (i,j) with i<=j
n_cross = len(i_triu)                      # = 10 for n_freq=4

print(f'Ell bins: {ell_bins.astype(int)}')
print(f'Number of unique freq-freq cross-spectra: {n_cross}')
print(f'Total data vector length: {n_ell_bins * n_cross} elements')

def compute_signal_model(r, A_lens, beta_d, A_d, alpha_d, beta_s, A_s, alpha_s):
    """
    Compute the multi-frequency BB signal model.

    Returns C_ell model array of shape (n_ell_bins, n_cross),
    where n_cross = n_freq*(n_freq+1)/2 = 10 (upper triangle of freq matrix).
    """
    # --- CMB power spectrum (same at all frequencies in K_CMB) ---
    Cl_cmb = cmb_Cl_BB(ell_bins, r=r, A_lens=A_lens)      # (n_ell_bins,)

    # --- Foreground SEDs at each band ---
    f_d = dust_sed(freqs, beta_d=beta_d)                   # (n_freq,)
    f_s = synch_sed(freqs, beta_s=beta_s)                  # (n_freq,)

    # --- Foreground power spectra ---
    Cl_dust  = Dl_to_Cl(dust_Dl_BB(ell_bins, A_d, alpha_d),  ell_bins)  # (n_ell_bins,)
    Cl_synch = Dl_to_Cl(synch_Dl_BB(ell_bins, A_s, alpha_s), ell_bins)  # (n_ell_bins,)

    # --- Build model: C_ell(nu_i, nu_j) = CMB + FG_dust + FG_synch ---
    model = np.zeros((n_ell_bins, n_cross))
    for icr, (i, j) in enumerate(zip(i_triu, j_triu)):
        model[:, icr] = (Cl_cmb
                         + f_d[i] * f_d[j] * Cl_dust
                         + f_s[i] * f_s[j] * Cl_synch)
    return model
```

```python
# ============================================================
# 4.2  Instrument noise model
# ============================================================

# Noise parameters: (sigma_pol [muK arcmin], FWHM_beam [arcmin])
# 93, 145 GHz: SO SAT-like;  217, 353 GHz: Planck-like
noise_params = {
     93: (6.0,  26.),
    145: (6.0,  17.),
    217: (50.,   5.),
    353: (500.,  5.),
}

def noise_Cl(ell, nu_GHz):
    """
    White noise C_ell [muK^2] with Gaussian beam deconvolution.
    N_ell = (sigma_pol)^2 * exp(ell*(ell+1) * sigma_b^2),
    where sigma_pol = sigma_arcmin * pi/(60*180) [muK rad].
    """
    sigma_arcmin, fwhm_arcmin = noise_params[int(round(nu_GHz))]
    sigma_rad = sigma_arcmin * np.pi / (60. * 180.)
    N0 = sigma_rad**2
    sigma_b = fwhm_arcmin * np.pi / (60. * 180.) / np.sqrt(8. * np.log(2.))
    return N0 * np.exp(ell * (ell + 1.) * sigma_b**2)

# Precompute noise at each ell bin for each band
N_ell = np.zeros((n_ell_bins, n_freq))
for ifreq, nu in enumerate(freqs):
    N_ell[:, ifreq] = noise_Cl(ell_bins, nu)

# Compare noise to signal at each band
print('Noise N_ell at ell=80 vs lensing C_ell at ell=80:')
il80 = np.argmin(np.abs(ell_bins - 80))
Cl_lens_80 = cmb_Cl_BB(80, r=0, A_lens=1.)
for ifreq, nu in enumerate(freqs):
    ratio = N_ell[il80, ifreq] / Cl_lens_80
    print(f'  {nu:.0f} GHz: N_ell/C_lens = {ratio:.1f}')
```

```python
# ============================================================
# 4.3  Generate synthetic data using the Knox covariance
# ============================================================

# True parameter values
params_true = dict(
    r=0.01, A_lens=1.0,
    beta_d=1.54, A_d=20., alpha_d=-0.4,
    beta_s=-3.0, A_s=2.,  alpha_s=-0.8,
)
f_sky = 0.10   # sky fraction (SO SAT high-latitude mask)

# True signal model
model_true = compute_signal_model(**params_true)    # (n_ell_bins, n_cross)

# Total C_ell matrix (signal + noise) for covariance calculation
def total_Cl_matrix(signal_model, N_ell_array):
    """
    Build full (n_ell_bins, n_freq, n_freq) total C_ell matrix
    from signal upper-triangle model and diagonal noise.
    """
    C_tot = np.zeros((n_ell_bins, n_freq, n_freq))
    for icr, (i, j) in enumerate(zip(i_triu, j_triu)):
        C_tot[:, i, j] = signal_model[:, icr]
        C_tot[:, j, i] = signal_model[:, icr]
    for ifreq in range(n_freq):
        C_tot[:, ifreq, ifreq] += N_ell_array[:, ifreq]
    return C_tot

def knox_covariance(C_tot, ell_bins, delta_ell, f_sky):
    """
    Knox covariance for all auto/cross frequency spectra.

    C_tot shape: (n_ell_bins, n_freq, n_freq)
    Returns covariance matrix with shape:
        (n_ell_bins * n_cross, n_ell_bins * n_cross)
    where n_cross = n_freq * (n_freq + 1) // 2.
    """

    n_ell_bins = len(ell_bins)
    n_freq = C_tot.shape[1]

    # Important: define these locally from C_tot, not from global variables
    i_triu, j_triu = np.triu_indices(n_freq)
    n_cross = len(i_triu)

    cov = np.zeros((n_ell_bins * n_cross, n_ell_bins * n_cross))

    for il, ell in enumerate(ell_bins):
        N_modes = (2 * ell + 1) * delta_ell * f_sky
        C = C_tot[il]

        for ia, (i, j) in enumerate(zip(i_triu, j_triu)):
            for ib, (k, l) in enumerate(zip(i_triu, j_triu)):

                row = il * n_cross + ia
                col = il * n_cross + ib

                cov[row, col] = (
                    C[i, k] * C[j, l] + C[i, l] * C[j, k]
                ) / N_modes

    return cov

C_tot_true = total_Cl_matrix(model_true, N_ell)
cov = knox_covariance(C_tot_true, ell_bins, delta_ell, f_sky)

# Invert covariance (used in likelihood)
inv_cov = np.linalg.solve(cov, np.eye(len(cov)))

# Generate a Gaussian realization of bandpowers around the true model
np.random.seed(42)
L = np.linalg.cholesky(cov)
data_vec = model_true.flatten() + L @ np.random.randn(n_ell_bins * n_cross)
data = data_vec.reshape((n_ell_bins, n_cross))   # (n_ell_bins, n_cross)

print(f'Data vector shape: {data.shape}  ({n_ell_bins} ell bins x {n_cross} freq pairs)')
print(f'Covariance matrix shape: {cov.shape}')
```

```python
# ============================================================
# Plotting code for synthetic BB bandpowers
# ============================================================

# Error bars from covariance
sigma = np.sqrt(np.diag(cov)).reshape((n_ell_bins, n_cross))


def get_cross_index(fi, fj):
    """
    Return the index of the frequency cross-spectrum fi x fj
    in the upper-triangular list of spectra.
    """
    for k, (a, b) in enumerate(zip(i_triu, j_triu)):
        if a == fi and b == fj:
            return k
    raise ValueError(f"Could not find cross-spectrum {fi} x {fj}")


def plot_one_spectrum(ax, fi, fj):
    """
    Plot one BB spectrum: data, CMB, dust, synchrotron, and total model.
    
    fi, fj are frequency indices.
    For example:
        fi = 0, fj = 0  means 93 x 93 GHz
        fi = 0, fj = 1  means 93 x 145 GHz
    """

    # --------------------------------------------------------
    # 1. Find the cross-spectrum index
    # --------------------------------------------------------
    icr = get_cross_index(fi, fj)

    nu_i = freqs[fi]
    nu_j = freqs[fj]
    label = f"{nu_i} x {nu_j} GHz"

    # --------------------------------------------------------
    # 2. Plot the synthetic data
    # --------------------------------------------------------
    ax.errorbar(
        ell_bins,
        Cl_to_Dl(data[:, icr], ell_bins),
        yerr=Cl_to_Dl(sigma[:, icr], ell_bins),
        fmt="o",
        color="k",
        ms=4,
        label="Data"
    )

    # --------------------------------------------------------
    # 3. Compute the CMB contribution
    # --------------------------------------------------------
    Cl_cmb = cmb_Cl_BB(
        ell_bins,
        r=params_true["r"],
        A_lens=params_true["A_lens"]
    )

    # --------------------------------------------------------
    # 4. Compute the dust contribution
    # --------------------------------------------------------
    dust_i = dust_sed(nu_i, beta_d=params_true["beta_d"])
    dust_j = dust_sed(nu_j, beta_d=params_true["beta_d"])

    Cl_dust_ref = Dl_to_Cl(
        dust_Dl_BB(
            ell_bins,
            params_true["A_d"],
            params_true["alpha_d"]
        ),
        ell_bins
    )

    Cl_dust = dust_i * dust_j * Cl_dust_ref

    # --------------------------------------------------------
    # 5. Compute the synchrotron contribution
    # --------------------------------------------------------
    sync_i = synch_sed(nu_i, beta_s=params_true["beta_s"])
    sync_j = synch_sed(nu_j, beta_s=params_true["beta_s"])

    Cl_sync_ref = Dl_to_Cl(
        synch_Dl_BB(
            ell_bins,
            params_true["A_s"],
            params_true["alpha_s"]
        ),
        ell_bins
    )

    Cl_sync = sync_i * sync_j * Cl_sync_ref

    # --------------------------------------------------------
    # 6. Plot the model components
    # --------------------------------------------------------
    ax.semilogy(
        ell_bins,
        Cl_to_Dl(Cl_cmb, ell_bins),
        "k--",
        lw=1.5,
        label="CMB"
    )

    ax.semilogy(
        ell_bins,
        Cl_to_Dl(Cl_dust, ell_bins),
        "C1:",
        lw=1.5,
        label="Dust"
    )

    ax.semilogy(
        ell_bins,
        Cl_to_Dl(Cl_sync, ell_bins),
        "C0:",
        lw=1.5,
        label="Synchrotron"
    )

    ax.semilogy(
        ell_bins,
        Cl_to_Dl(model_true[:, icr], ell_bins),
        "C2-",
        lw=2,
        label="Total model"
    )

    # --------------------------------------------------------
    # 7. Labels and formatting
    # --------------------------------------------------------
    ax.set_title(label)
    ax.set_xlabel(r"$\ell$")
    ax.set_ylabel(r"$D_\ell^{BB}$ [$\mu K^2$]")
    ax.set_xlim(20, 320)


# ============================================================
# plot only 93 x 93 GHz
# ============================================================

fig, ax = plt.subplots(figsize=(6, 4.5))

plot_one_spectrum(ax, fi=0, fj=0)

ax.legend(fontsize=8)
ax.set_title("93 x 93 GHz")

plt.tight_layout()
plt.show()
```

## Section 5: The Likelihood

### From data to parameters

We now have data $\mathbf{d}$ (90 bandpowers) and a model $\mathbf{m}(\theta)$ with 8 parameters. The **likelihood** answers the question: *how probable is it to measure $\mathbf{d}$ if the true parameters were $\theta$?* Because our bandpowers are Gaussian-distributed (that is how we generated them in Section 4), the answer is the familiar chi-squared:

$$-2\ln \mathcal{L}(\theta) = \chi^2(\theta) = \left[\mathbf{d} - \mathbf{m}(\theta)\right]^T \mathbf{C}^{-1} \left[\mathbf{d} - \mathbf{m}(\theta)\right],$$

with $\mathbf{C}$ the Knox covariance from Section 4, held fixed at the fiducial parameters — a standard approximation, valid when the covariance varies slowly with the parameters. (Real analyses at low signal-to-noise often upgrade to the Hamimeche–Lewis likelihood; see Exercise 6.)

### Priors

Bayes' theorem, $P(\theta \,|\, \mathbf{d}) \propto \mathcal{L}(\mathbf{d} \,|\, \theta)\, \pi(\theta)$, needs one more ingredient: the **prior** $\pi(\theta)$ — what we knew before this measurement. We use wide flat (uniform) priors on the amplitudes and slopes, and Gaussian priors on the two spectral indices, representing external knowledge (e.g. from Planck):

| Parameter | Symbol | Physical meaning | Prior |
|:----------|:-------|:----------------|:------|
| Tensor-to-scalar ratio | $r$ | Amplitude of primordial GWs | $U[0,\, 0.5]$ |
| Lensing amplitude | $A_{\rm lens}$ | Lensing B-mode amplitude | $U[0,\, 2]$ |
| Dust spectral index | $\beta_d$ | MBB slope | $G(1.54,\, 0.1)$ |
| Dust amplitude | $A_d$ | BB amplitude at $\ell=80$, 353 GHz | $U[0,\, 200]$ |
| Dust slope | $\alpha_d$ | Power law in $\ell$ | $U[-1,\, 0]$ |
| Synch. spectral index | $\beta_s$ | Power law slope | $G(-3.0,\, 0.3)$ |
| Synch. amplitude | $A_s$ | BB amplitude at $\ell=80$, 23 GHz | $U[0,\, 50]$ |
| Synch. slope | $\alpha_s$ | Power law in $\ell$ | $U[-2,\, 0]$ |


```python
# ============================================================
# 5.1  Likelihood, prior, and posterior
# ============================================================

param_names = ['r', 'A_lens', 'beta_d', 'A_d', 'alpha_d', 'beta_s', 'A_s', 'alpha_s']
p0_true     = np.array([0.01, 1.0, 1.54, 20., -0.4, -3.0, 2., -0.8])

def log_prior(params_vec):
    """Flat priors within the bounds of the table above,
    plus Gaussian priors on the two spectral indices."""
    r, A_lens, beta_d, A_d, alpha_d, beta_s, A_s, alpha_s = params_vec
    if not (0 <= r <= 0.5 and 0 <= A_lens <= 2 and 0 <= A_d <= 200 and
            -1 <= alpha_d <= 0 and 0 <= A_s <= 50 and -2 <= alpha_s <= 0):
        return -np.inf                                   # outside a flat prior
    return (-0.5 * ((beta_d - 1.54) / 0.1) ** 2          # Gaussian on beta_d
            - 0.5 * ((beta_s + 3.0)  / 0.3) ** 2)        # Gaussian on beta_s

def log_likelihood(params_vec):
    """Gaussian log-likelihood: -chi^2 / 2."""
    model = compute_signal_model(**dict(zip(param_names, params_vec)))
    dx = (data - model).flatten()
    return -0.5 * dx @ inv_cov @ dx

def log_posterior(params_vec):
    """Bayes: log-posterior = log-prior + log-likelihood (up to a constant)."""
    lp = log_prior(params_vec)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_likelihood(params_vec)

# Sanity check at the true parameters: chi^2 per degree of freedom ~ 1
chi2_true = -2 * log_likelihood(p0_true)
n_dof = n_ell_bins * n_cross - len(param_names)
print(f'chi^2 at true parameters: {chi2_true:.1f}  (N_dof = {n_dof})')
print(f'Reduced chi^2: {chi2_true / n_dof:.2f}  (expect ~1.0 for a good fit)')
```

```python
# ============================================================
# 5.2  1D profile likelihood scan over r
# ============================================================

# Fix all other parameters at truth and scan over r
r_arr   = np.linspace(0., 0.15, 50)
chi2_1d = np.zeros(len(r_arr))

for ir, r_val in enumerate(r_arr):
    p_test = p0_true.copy()
    p_test[0] = r_val
    chi2_1d[ir] = -2 * log_likelihood(p_test)

fig, ax = plt.subplots(figsize=(5, 4))
ax.plot(r_arr, chi2_1d - chi2_1d.min(), 'C2-', lw=2.5)
ax.axvline(params_true['r'], color='k', ls='--', label=f"True $r={params_true['r']}$")
ax.axhline(1.0, color='grey', ls=':', label=r'$\Delta\chi^2=1$ (1$\sigma$)')
ax.axhline(4.0, color='grey', ls='--', label=r'$\Delta\chi^2=4$ (2$\sigma$)')
ax.set_xlabel('$r$')
ax.set_ylabel(r'$\Delta\chi^2$')
ax.set_title('Profile likelihood for $r$ (all other parameters fixed at truth)')
ax.legend()
ax.set_xlim(0, 0.06)
ax.set_ylim(-0.5, 10)
plt.tight_layout()
plt.show()
```

**Key insight**: This is a 1D slice: in reality we need to marginalize over all foregrounds parameters. The true posterior on r will be broader due to foreground parameter degeneracies.

## Section 6: Parameter Estimation with Markov Chain Monte Carlo (MCMC)

### Why we can't just use a grid

What we want to report is the **marginalized** posterior of $r$: the 8-dimensional posterior $P(\theta|\mathbf{d})$ integrated over the 7 nuisance parameters. On a grid with a modest 50 points per axis, that is $50^8 \approx 4\times10^{13}$ likelihood evaluations — centuries of computing. Beyond 3–4 dimensions, grids are hopeless.

### The idea

MCMC replaces the grid with a **random walk that spends its time in proportion to the posterior**: where the posterior is twice as high, the chain leaves twice as many samples. Once you have such samples, everything becomes easy:

- marginalizing = simply *ignoring the other columns* of the chain;
- the histogram of the $r$ column **is** $p(r|\mathbf{d})$;
- medians and credible intervals are ordinary sample statistics.

### The Metropolis–Hastings recipe

1. Start somewhere: $\theta_0$.
2. **Propose** a random step: $\theta' = \theta_t + \Delta\theta$, with $\Delta\theta$ Gaussian with fixed widths.
3. Compute the posterior ratio $a = P(\theta'|\mathbf{d}) / P(\theta_t|\mathbf{d})$ (in logs, in practice).
4. **Accept** with probability $\min(1, a)$: if accepted, $\theta_{t+1} = \theta'$; if rejected, $\theta_{t+1} = \theta_t$ — *repeat the current point in the chain* (essential, don't skip it!).
5. Go to 2.

Uphill moves are always accepted; downhill moves sometimes — that is what lets the chain trace out the full width of the posterior rather than climbing to the peak and staying there. And since only posterior *ratios* appear, we never need the normalization.

Three practical rules of thumb:

- **Step size**: aim for roughly 20–40% of proposals accepted (too big → everything rejected; too small → the chain crawls).
- **Burn-in**: throw away the start of the chain, which remembers the arbitrary starting point.
- **Convergence**: run several chains from different starting points and check that they agree (the Gelman–Rubin statistic $\hat{R} \approx 1$).

One prediction before we run: the posterior on $r$ will come out **broader** than the 1D slice of Section 5.2. That broadening is the price of marginalizing over the unknown foregrounds — and the reason multi-frequency coverage matters.

Define the sampler:

```python
# ============================================================
# 6.1  A Metropolis-Hastings sampler from scratch
# ============================================================
# The algorithm can be written in ~15 lines of python!

def run_metropolis_hastings(log_post, p_start, prop_sigma, n_steps, seed=0):
    """
    Metropolis-Hastings sampler with an independent Gaussian proposal
    of width prop_sigma[i] in each parameter.

    Returns
    -------
    chain       : (n_steps, ndim) array of posterior samples
    accept_frac : fraction of proposed moves that were accepted
    """
    rng    = np.random.default_rng(seed)
    ndim   = len(p_start)
    chain  = np.empty((n_steps, ndim))
    p_cur  = np.array(p_start, dtype=float)
    lp_cur = log_post(p_cur)
    n_accept = 0

    for i in range(n_steps):
        # 1. propose a move
        p_prop  = p_cur + prop_sigma * rng.standard_normal(ndim)
        lp_prop = log_post(p_prop)

        # 2. Metropolis acceptance rule (in log space)
        if np.log(rng.random()) < lp_prop - lp_cur:
            p_cur, lp_cur = p_prop, lp_prop
            n_accept += 1

        # 3. store the current point (repeated if the move was rejected!)
        chain[i] = p_cur

    return chain, n_accept / n_steps
```

### Seeing the algorithm in action

Before unleashing the sampler on the full 8-dimensional problem, let's *watch* it work on a 2-dimensional slice of our posterior: $(r, A_d)$, with the other six parameters held fixed at their true values. In 2D we can still afford a brute-force grid evaluation of the posterior (impossible in 8D!), so we can draw the exact answer the chain is supposed to reproduce. Three things to look for:

1. **Burn-in**: we start the chain far from the peak on purpose. The first steps drift "uphill" and clearly remember the starting point — this is the part we discard.
2. **Rejections are part of the algorithm**: red crosses are proposed moves that were rejected; after each one the chain *repeats* its current position. This repetition is what keeps the sample density proportional to the posterior.
3. **The histogram converges to the posterior**: given enough steps, the density of chain samples reproduces the exact (grid-computed) posterior, and a 1D histogram of the samples reproduces the exact 1D marginal.

```python
# ============================================================
# 6.2  Watching the random walk explore a 2D posterior
# ============================================================

def log_post_2d(r_val, A_d_val):
    """Posterior in the (r, A_d) plane, other parameters fixed at truth."""
    p = p0_true.copy()
    p[0], p[3] = r_val, A_d_val
    return log_posterior(p)

# --- Exact posterior on a grid (only affordable because this is 2D) ---
r_grid  = np.linspace(0., 0.055, 80)
Ad_grid = np.linspace(16., 22.5, 80)
lnP_grid = np.array([[log_post_2d(rv, av) for rv in r_grid] for av in Ad_grid])
P_grid = np.exp(lnP_grid - lnP_grid.max())
# contour levels enclosing 1, 2, 3 sigma for a 2D Gaussian
sigma_levels = [np.exp(-11.8 / 2), np.exp(-6.17 / 2), np.exp(-2.30 / 2)]

# --- An instrumented M-H run that also records rejected proposals ---
rng_viz = np.random.default_rng(4)
n_view  = 5000
prop2d  = np.array([0.007, 0.6])        # ~ conditional posterior widths
p_cur   = np.array([0.050, 21.8])       # deliberately bad starting point
lp_cur  = log_post_2d(*p_cur)
path         = np.empty((n_view + 1, 2)); path[0] = p_cur
proposals    = np.empty((n_view, 2))
was_accepted = np.empty(n_view, dtype=bool)
for k in range(n_view):
    p_new  = p_cur + prop2d * rng_viz.standard_normal(2)
    lp_new = log_post_2d(*p_new)
    proposals[k] = p_new
    was_accepted[k] = np.log(rng_viz.random()) < lp_new - lp_cur
    if was_accepted[k]:
        p_cur, lp_cur = p_new, lp_new
    path[k + 1] = p_cur
print(f'2D chain: acceptance fraction = {was_accepted.mean():.2f}')

# --- Three-panel figure ---
fig, axes = plt.subplots(1, 3, figsize=(15, 4.6))

# Panel 1: the first steps, proposal by proposal
n1 = 150
ax = axes[0]
ax.contour(r_grid, Ad_grid, P_grid, levels=sigma_levels,
           colors='k', linewidths=1, alpha=0.5)
ok = was_accepted[:n1]
ax.plot(path[:n1 + 1, 0], path[:n1 + 1, 1], '-', color='C0', lw=1, alpha=0.9)
ax.plot(proposals[:n1][~ok, 0], proposals[:n1][~ok, 1], 'x', color='C3',
        ms=5, mew=1.5, label='rejected proposals')
ax.plot(proposals[:n1][ok, 0], proposals[:n1][ok, 1], 'o', color='C2',
        ms=3.5, label='accepted proposals')
ax.plot(*path[0], '*', color='k', ms=16, label='start')
ax.set_xlabel('$r$'); ax.set_ylabel('$A_d$')
ax.set_title(f'First {n1} steps: burn-in')
ax.legend(fontsize=8, loc='lower right')

# Panel 2: the full chain maps out the posterior
ax = axes[1]
ax.plot(path[:, 0], path[:, 1], '.', color='C0', ms=1.5, alpha=0.15)
ax.contour(r_grid, Ad_grid, P_grid, levels=sigma_levels,
           colors='k', linewidths=1.2)
ax.set_xlabel('$r$'); ax.set_ylabel('$A_d$')
ax.set_title(f'All {n_view} samples vs exact contours')
ax.set_xlim(r_grid[0], r_grid[-1]); ax.set_ylim(Ad_grid[0], Ad_grid[-1])

# Panel 3: histogram of the samples = the exact marginal posterior
ax = axes[2]
p_r_exact = P_grid.sum(axis=0)
p_r_exact /= np.trapz(p_r_exact, r_grid)
n_burn_2d = 500
ax.hist(path[n_burn_2d:, 0], bins=40, density=True, alpha=0.6, color='C0',
        label=f'chain histogram\n(after {n_burn_2d} burn-in steps)')
ax.plot(r_grid, p_r_exact, 'k-', lw=2, label='exact posterior (grid)')
ax.axvline(params_true['r'], color='r', ls='--', label='true $r$')
ax.set_xlabel('$r$'); ax.set_ylabel('Posterior density')
ax.set_title('Histogram of samples = posterior')
ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

**Key insight**: The chain never computes the normalization of the posterior, and never evaluates it on a grid - yet its samples reproduce it exactly.

```python
# ============================================================
# 6.3  Run several chains from dispersed starting points
# ============================================================
import time

ndim     = len(param_names)
n_chains = 4       # independent chains, for convergence checks
n_steps  = 25000   # steps per chain
n_burn   = 5000    # burn-in steps (discarded)

# Proposal widths: a rough guess of each parameter's posterior width,
# times an overall factor tuned so that ~25% of proposals get accepted.
# (Exercise 5 explores what happens when this is badly tuned.)
sigma_guess = np.array([0.008, 0.06, 0.03, 1.0, 0.06, 0.3, 2.0, 0.6])
prop_sigma  = 0.45 * sigma_guess

# Dispersed starting points around the fiducial parameters
rng_init = np.random.default_rng(1)
p_spread = np.array([5e-3, 0.1, 0.05, 4., 0.1, 0.1, 0.5, 0.2])

t0 = time.time()
chains = []
for ic in range(n_chains):
    p_start = p0_true + p_spread * rng_init.standard_normal(ndim)
    p_start[0] = abs(p_start[0])   # keep r inside its prior
    chain, acc = run_metropolis_hastings(log_posterior, p_start, prop_sigma,
                                         n_steps, seed=100 + ic)
    chains.append(chain)
    print(f'Chain {ic}: acceptance fraction = {acc:.2f}  (target ~0.2-0.4)')
chains = np.array(chains)          # shape (n_chains, n_steps, ndim)
print(f'Done in {time.time()-t0:.1f} s '
      f'({n_chains * n_steps} likelihood evaluations)')

# Discard burn-in and merge the chains into one set of posterior samples
flat_samples = chains[:, n_burn:, :].reshape(-1, ndim)
print(f'Posterior samples after burn-in: {flat_samples.shape[0]}')
```

```python
# ============================================================
# 6.4  Convergence diagnostics: trace plots
# ============================================================

fig, axes = plt.subplots(4, 2, figsize=(13, 10), sharex=True)
labels_plot = [r'$r$', r'$A_{\rm lens}$', r'$\beta_d$', r'$A_d$',
               r'$\alpha_d$', r'$\beta_s$', r'$A_s$', r'$\alpha_s$']

for i, (ax, label) in enumerate(zip(axes.flatten(), labels_plot)):
    for ic in range(n_chains):
        ax.plot(chains[ic, :, i], alpha=0.5, lw=0.5)
    ax.axhline(p0_true[i], color='r', ls='--', lw=1.5, label='True')
    ax.axvline(n_burn, color='k', ls=':', lw=1.5, label='End burn-in')
    ax.set_ylabel(label)
    if i == 0:
        ax.legend(fontsize=9)

axes[-1, 0].set_xlabel('Step')
axes[-1, 1].set_xlabel('Step')
plt.suptitle('MCMC trace plots (one color per chain; dotted line = end of burn-in)',
             y=1.01)
plt.tight_layout()
plt.show()

print('All chains should look statistically identical after burn-in,')
print('regardless of where they started. The next cell turns this')
print('visual check into a number.')
```

```python
# ============================================================
# 6.5  Convergence check: the Gelman-Rubin statistic R-hat
# ============================================================

def gelman_rubin(chains_burned):
    """
    R-hat convergence diagnostic (Gelman & Rubin 1992).

    chains_burned: (n_chains, n_samples, ndim), burn-in already removed.
    Compares the between-chain variance B to the within-chain variance W;
    values close to 1 (conventionally < 1.05) indicate convergence.
    """
    m, n, _ = chains_burned.shape
    chain_means = chains_burned.mean(axis=1)           # (m, ndim)
    chain_vars  = chains_burned.var(axis=1, ddof=1)    # (m, ndim)
    W = chain_vars.mean(axis=0)                        # within-chain variance
    B = n * chain_means.var(axis=0, ddof=1)            # between-chain variance
    var_est = (n - 1) / n * W + B / n                  # pooled estimate
    return np.sqrt(var_est / W)

rhat = gelman_rubin(chains[:, n_burn:, :])
print(f'{"Parameter":<10} {"R-hat":>8}')
print('-' * 20)
for name, r_ in zip(param_names, rhat):
    flag = '' if r_ < 1.05 else '   <-- not converged: run longer!'
    print(f'{name:<10} {r_:>8.3f}{flag}')
```

```python
# ============================================================
# 6.6  Posterior distributions: 1D marginals
# ============================================================

fig, axes = plt.subplots(2, 4, figsize=(14, 6))

for i, (ax, label) in enumerate(zip(axes.flatten(), labels_plot)):
    ax.hist(flat_samples[:, i], bins=40, density=True,
            color='C0', alpha=0.7, edgecolor='white', lw=0.5)
    ax.axvline(p0_true[i], color='r', ls='--', lw=2, label='True')

    # 1-sigma credible interval
    lo, mid, hi = np.percentile(flat_samples[:, i], [16, 50, 84])
    ax.axvline(lo,  color='C2', ls=':',  lw=1.5)
    ax.axvline(hi,  color='C2', ls=':',  lw=1.5)
    ax.axvline(mid, color='C2', ls='-',  lw=2)

    ax.set_xlabel(label)
    ax.set_yticks([])
    ax.set_title(f'{label} = ${mid:.3g}^{{+{hi-mid:.3g}}}_{{-{mid-lo:.3g}}}$', fontsize=9)
    if i == 0:
        ax.legend(fontsize=9)

plt.suptitle('Marginalized 1D posteriors (green: median \u00b1 1\u03c3; red: truth)', y=1.01)
plt.tight_layout()
plt.show()

# Summary table
print(f'{"Parameter":<12}  {"Truth":>8}  {"Median":>10}  {"68% CI":>15}')
print('-' * 52)
for i, name in enumerate(param_names):
    lo, mid, hi = np.percentile(flat_samples[:, i], [16, 50, 84])
    print(f'{name:<12}  {p0_true[i]:>8.4g}  {mid:>10.4g}  +{hi-mid:.3g} / -{mid-lo:.3g}')
```

### From this toy pipeline to a real analysis

Production component-separation pipelines extend what we built here in several important ways, e.g.:

1. **Bandpass integration**: instead of evaluating SEDs at a single frequency, real pipelines integrate them over the measured bandpass profile $b(\nu)$ of each channel, accounting for the finite width of every frequency band.

2. **Bandpower window functions**: proper window functions $W_b(\ell)$ replace our simple bin averages; they encode the exact mode coupling induced by sky masking (see the [Power Spectrum Analysis notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Power_Spectrum_Analysis.ipynb)).

5. **Foreground complexity**: real foregrounds vary across the sky. This needs to be modelled to avoid biased cosmological constraints.

6. **Covariance from simulations**: the bandpower covariance is usually estimated from large ensembles of realistic map simulations rather than from the analytic Knox formula.

The characteristic primordial $B$-mode polarization pattern peaks on degree angular scales — the “recombination bump” at $\ell \approx 80$ — with an additional “reionization bump” on very large angular scales, $\ell \lesssim 10$. The difficulty is that we do not know the value of $r$ to expect: current observations only give an upper limit, not a prediction. If $r$ is close to the current limit, the signal may be within reach; if it is much smaller, it could be far harder to detect.

For example, for $r=0.01$, the primordial signal is already below the lensing $B$-mode signal over much of the range $\ell \gtrsim 80$. In this sense, lensing $B$-modes act like a foreground for primordial $B$-mode searches: they are a real cosmological signal, but one that obscures the even fainter signal from primordial gravitational waves.

The value of (r) is highly model dependent. In single-field slow-roll inflation, (r) is related to the steepness of the inflationary potential through (r\simeq 16\epsilon). Large-field models generally predict larger tensor signals, while plateau-like models such as Starobinsky- or Higgs-like inflation predict much smaller values, often near the (10^{-3}) scale. Other models can make (r) arbitrarily small. Therefore, current upper limits do not tell us what (r) “should” be; they tell us which inflationary models are already under pressure. Future experiments are exciting because they will either detect (r), pointing toward a high-energy, gravitational-wave-producing early Universe, or push many large-field models out of the viable parameter space.

A neat trick: we do not need to regenerate the data. The 2-band data vector is just a *sub-selection* of the 4-band one (keep the 93×93, 93×145 and 145×145 pairs in every $\ell$ bin), and its covariance is the matching sub-block of the full covariance.

```python
# ============================================================
# A.1  Personal solution - Exercise 1: only 93 + 145 GHz
# ============================================================

def sig68(samples_1d):
    """Half the 16th-84th percentile range: a robust 1-sigma estimate."""
    lo, hi = np.percentile(samples_1d, [16, 84])
    return 0.5 * (hi - lo)

def quick_mcmc(log_post, p_center=None, prop_scale=1.0,
               n_chains_q=2, n_steps_q=15000, n_burn_q=3000, seed0=200):
    """A couple of short MH chains + acceptance / R-hat printout."""
    if p_center is None:
        p_center = p0_true
    rng_q = np.random.default_rng(seed0)
    ch, accs = [], []
    for ic in range(n_chains_q):
        p_start = p_center + p_spread * rng_q.standard_normal(ndim)
        p_start[0] = abs(p_start[0])                    # r >= 0
        p_start[1] = np.clip(p_start[1], 0.01, 1.99)    # A_lens in prior
        c_, acc = run_metropolis_hastings(log_post, p_start,
                                          prop_scale * prop_sigma,
                                          n_steps_q, seed=seed0 + 10 + ic)
        ch.append(c_[n_burn_q:])
        accs.append(acc)
    ch = np.array(ch)
    print(f'  acceptance = {np.mean(accs):.2f}, '
          f'max R-hat = {gelman_rubin(ch).max():.3f}')
    return ch.reshape(-1, ndim)

def subset_log_posterior(band_idx, log_prior_fn=None):
    """Posterior for a subset of the frequency bands, built by
    sub-selecting the 4-band data vector and covariance."""
    if log_prior_fn is None:
        log_prior_fn = log_prior
    band_idx = list(band_idx)
    pair_sel = np.array([icr for icr, (i, j) in enumerate(zip(i_triu, j_triu))
                         if i in band_idx and j in band_idx])
    rows = np.concatenate([il * n_cross + pair_sel for il in range(n_ell_bins)])
    data_sub    = data_vec[rows]
    inv_cov_sub = np.linalg.inv(cov[np.ix_(rows, rows)])

    freqs_sub = freqs[band_idx]
    ii, jj = np.triu_indices(len(band_idx))

    def log_post_sub(pv):
        lp = log_prior_fn(pv)
        if not np.isfinite(lp):
            return -np.inf
        r, A_lens, beta_d, A_d, alpha_d, beta_s, A_s, alpha_s = pv
        Cl_cmb = cmb_Cl_BB(ell_bins, r=r, A_lens=A_lens)
        f_d = dust_sed(freqs_sub, beta_d=beta_d)
        f_s = synch_sed(freqs_sub, beta_s=beta_s)
        Cl_d = Dl_to_Cl(dust_Dl_BB(ell_bins, A_d, alpha_d), ell_bins)
        Cl_s = Dl_to_Cl(synch_Dl_BB(ell_bins, A_s, alpha_s), ell_bins)
        m = np.stack([Cl_cmb + f_d[i] * f_d[j] * Cl_d + f_s[i] * f_s[j] * Cl_s
                      for i, j in zip(ii, jj)], axis=1)
        dx = data_sub - m.flatten()
        return lp - 0.5 * dx @ inv_cov_sub @ dx
    return log_post_sub

print('2-band (93+145) run:')
samples_2band = quick_mcmc(subset_log_posterior([0, 1]),
                           n_steps_q=20000, n_burn_q=4000)

print(f'\nsigma(r) with 4 bands: {sig68(flat_samples[:, 0]):.4f}')
print(f'sigma(r) with 2 bands: {sig68(samples_2band[:, 0]):.4f}')
print('\nWithout the dust-monitoring high frequencies, the dust left in the')
print('CMB channels cannot be separated from the CMB, and sigma(r) inflates.')
```

### Exercise 2 — Delensing *(personal solution)*

Lensing B-modes act as an extra source of *sample variance*: even though $A_\mathrm{lens}$ is measured well, the lensing power contributes to the covariance of every bandpower ($\propto C_\ell^\mathrm{tot}$ in the Knox formula). Reducing $A_\mathrm{lens}$ in the data (which is what delensing does — see the [Lensing Reconstruction notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Lensing_Reconstruction.ipynb)) shrinks that variance and tightens $\sigma(r)$. Note that in our setup the CMB channels are close to noise-dominated ($N_\ell \sim 2\,C_\ell^\mathrm{lens}$ at $\ell \sim 80$), so the improvement is real but not dramatic; for a deeper experiment, delensing matters more.

```python
# ============================================================
# A.2  Personal solution - Exercise 2: delensing
# ============================================================

def sigma_r_delensed(A_lens_true, seed=3):
    """Regenerate data with a delensed A_lens and re-fit."""
    params_mod = dict(params_true, A_lens=A_lens_true)
    model_mod  = compute_signal_model(**params_mod)
    C_tot_mod  = total_Cl_matrix(model_mod, N_ell)
    cov_mod    = knox_covariance(C_tot_mod, ell_bins, delta_ell, f_sky)
    inv_cov_mod = np.linalg.inv(cov_mod)
    rng_mod = np.random.default_rng(seed)
    data_mod = model_mod.flatten() \
               + np.linalg.cholesky(cov_mod) @ rng_mod.standard_normal(len(cov_mod))

    def log_post_mod(pv):
        lp = log_prior(pv)
        if not np.isfinite(lp):
            return -np.inf
        dx = data_mod - compute_signal_model(**dict(zip(param_names, pv))).flatten()
        return lp - 0.5 * dx @ inv_cov_mod @ dx

    p_center = p0_true.copy()
    p_center[1] = max(A_lens_true, 0.05)   # start chains near the new truth
    samples_mod = quick_mcmc(log_post_mod, p_center=p_center)
    return sig68(samples_mod[:, 0])

print(f'{"A_lens in data":>15}  {"sigma(r)":>9}')
for A_lens_true in [1.0, 0.5, 0.0]:
    print(f'{A_lens_true:>15.1f}', end='  ')
    sig = sigma_r_delensed(A_lens_true)
    print(f'{"":>0}{sig:>9.4f}')
```

### Exercise 3 — Dust–CMB Degeneracy *(personal solution)*

At a *single* frequency, a bandpower excess can be attributed either to CMB ($r$) or to dust ($A_d$ scaled to that frequency): the two are perfectly degenerate and $r$ is unmeasurable. What breaks the degeneracy is the *frequency dependence*: dust power scales as $f_d(\nu_i)\,f_d(\nu_j)$ across frequency pairs while the CMB is flat.

With all 4 bands (including the 353 GHz dust monitor) the joint $(r, A_d)$ posterior is nearly uncorrelated: the dust-dominated channels measure $A_d$ on their own, independently of $r$. With only the 2 CMB channels of Exercise 1, a clear correlation appears and the contour area grows substantially ($\mathrm{corr} \approx +0.4$ in our realization; the *sign* of the correlation is not trivial to predict, because the fit can also trade power against the synchrotron component and is shaped by the $\beta_d$ prior — the robust lesson is the loss of the independent dust measurement, not the tilt direction).

```python
# ============================================================
# A.3  Personal solution - Exercise 3: the (r, A_d) posterior
# ============================================================
# (uses samples_2band from solution A.1)

fig, axes = plt.subplots(1, 2, figsize=(11.5, 4.6))
for ax, s, ttl in [(axes[0], flat_samples,  '4 bands'),
                   (axes[1], samples_2band, '2 bands (93+145), from A.1')]:
    ax.hist2d(s[:, 0], s[:, 3], bins=45, cmap='Blues')
    rho = np.corrcoef(s[:, 0], s[:, 3])[0, 1]
    ax.axvline(params_true['r'],  color='r', ls='--', lw=1)
    ax.axhline(params_true['A_d'], color='r', ls='--', lw=1)
    ax.set_xlabel('$r$'); ax.set_ylabel('$A_d$')
    ax.set_title(f'{ttl}:  corr$(r, A_d) = {rho:+.2f}$')
plt.tight_layout()
plt.show()

print('With 353 GHz in the fit, A_d is pinned by the dust-dominated channels')
print('and the (r, A_d) posterior decorrelates; with 2 bands it does not.')
```

### Exercise 4 — Dust Spectral Index Prior *(personal solution)*

Perhaps surprisingly, with all 4 bands the answer is: *almost nothing changes*. The 217 and 353 GHz channels measure $\beta_d$ from the data alone to $\sim \pm 0.025$ — much better than the $\pm 0.1$ Gaussian prior — so the prior is essentially irrelevant here. The prior becomes decisive exactly when the data cannot constrain $\beta_d$ internally: try re-running the 2-band setup of Exercise 1 with the flat prior and watch $\sigma(r)$ blow up. This is why external information (e.g. Planck 353 GHz dust maps) is so valuable for ground-based experiments with limited frequency coverage.

```python
# ============================================================
# A.4  Personal solution - Exercise 4: flat prior on beta_d
# ============================================================

def log_prior_flat_beta(params_vec):
    """Same as log_prior, but beta_d gets a flat prior U[1.0, 2.5]."""
    r, A_lens, beta_d, A_d, alpha_d, beta_s, A_s, alpha_s = params_vec
    if not (0 <= r <= 0.5 and 0 <= A_lens <= 2 and 1.0 <= beta_d <= 2.5 and
            0 <= A_d <= 200 and -1 <= alpha_d <= 0 and
            0 <= A_s <= 50 and -2 <= alpha_s <= 0):
        return -np.inf
    return -0.5 * ((beta_s + 3.0) / 0.3) ** 2   # keep the synchrotron prior

def log_post_flatbeta(pv):
    lp = log_prior_flat_beta(pv)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_likelihood(pv)

print('4-band run with flat prior on beta_d:')
samples_flatbeta = quick_mcmc(log_post_flatbeta)

print(f'\n{"":>24} {"Gaussian prior":>15} {"flat prior":>12}')
print(f'{"sigma(r)":>24} {sig68(flat_samples[:, 0]):>15.4f} '
      f'{sig68(samples_flatbeta[:, 0]):>12.4f}')
print(f'{"sigma(beta_d)":>24} {sig68(flat_samples[:, 2]):>15.4f} '
      f'{sig68(samples_flatbeta[:, 2]):>12.4f}')
print('\nThe 217+353 GHz channels measure beta_d better than the prior does,')
print('so the prior barely matters here - it matters when they are absent.')
```

### Exercise 5 — Tuning the MCMC Proposal *(personal solution)*

- **`prop_sigma` × 10**: almost every proposal lands far outside the posterior bulk and is rejected (acceptance $\sim$ 1%). The chain sits still for long stretches — the trace looks like a staircase — and yields only a handful of distinct samples.
- **`prop_sigma` × 0.1**: almost everything is accepted (acceptance $\sim$ 90%), which *sounds* good but is not: the steps are so small that the chain is a slowly drifting random walk that takes thousands of steps to cross the posterior once. Successive samples are extremely correlated.
- **`prop_sigma` × 1**: acceptance $\sim$ 25%, the chain decorrelates in tens of steps.

Both extremes are still formally correct MCMC — they would converge eventually — but the number of *effectively independent* samples per CPU-hour collapses. Acceptance fraction near 0 **or** near 1 is a warning sign.

```python
# ============================================================
# A.5  Personal solution - Exercise 5: proposal width extremes
# ============================================================

fig, axes = plt.subplots(3, 1, figsize=(9, 7.5), sharex=True)
for ax, scale in zip(axes, [10., 1., 0.1]):
    chain_t, acc = run_metropolis_hastings(log_posterior, p0_true,
                                           scale * prop_sigma, 4000, seed=11)
    ax.plot(chain_t[:, 0], lw=0.7, color='C0')
    ax.set_ylabel('$r$')
    ax.set_title(f'prop_sigma x {scale:g}:  acceptance = {acc:.3f}', fontsize=10)
axes[-1].set_xlabel('Step')
plt.suptitle('Trace of $r$ for three proposal widths (chains started at truth)')
plt.tight_layout()
plt.show()
```

### Exercise 6 — Hamimeche–Lewis vs Gaussian Likelihood *(personal solution)*

The [Hamimeche & Lewis (2008)](https://arxiv.org/abs/0801.0554) likelihood handles the fact that power spectrum estimates are not Gaussian-distributed (a sum of squares follows a Wishart distribution: positive, skewed). For each $\ell$ bin one forms the $N_\nu \times N_\nu$ matrices of measured ($\hat{C}_b$) and model ($C_b(\theta)$) bandpowers, applies the matrix function
$$\mathbf{X}_b = \mathbf{C}_{f,b}^{1/2}\; g\!\left(\mathbf{C}_b^{-1/2}\, \hat{\mathbf{C}}_b\, \mathbf{C}_b^{-1/2}\right) \mathbf{C}_{f,b}^{1/2}, \qquad g(x) = \mathrm{sign}(x-1)\sqrt{2\,(x - \ln x - 1)},$$
(with $g$ acting on the eigenvalues, and $\mathbf{C}_{f,b}$ a fixed fiducial model), and then feeds the vectorized $\mathbf{X}_b$ through the ordinary Gaussian covariance. Near the peak $g(x) \approx x - 1$, so the H&L likelihood reduces to the Gaussian one when the data are close to the model.

**Expected outcome here**: the two posteriors are nearly indistinguishable, for two reasons. First, our bins have $N_\mathrm{modes} \approx 10^3$ modes each, so the central limit theorem has already made the bandpowers very Gaussian. Second — an honesty caveat — our synthetic data were *generated* as Gaussian draws, so the Gaussian likelihood is exactly right for them by construction; this comparison is a consistency check, not a demonstration that H&L is better. The difference matters for real data at low $N_\mathrm{modes}$: small sky fractions, the lowest multipoles (e.g. the reionization bump at $\ell < 10$), or unbinned spectra.

```python
# ============================================================
# A.6  Personal solution - Exercise 6: H&L likelihood
# ============================================================

def mat_pow(M, p):
    """Power of a symmetric positive-definite matrix via eigen-decomposition."""
    w, V = np.linalg.eigh(M)
    return (V * w**p) @ V.T

def hl_g(x):
    return np.sign(x - 1.) * np.sqrt(2. * (x - np.log(x) - 1.))

# Fixed ingredients: data matrices (noise on the diagonal, since the
# theory matrix C_b must include noise to be positive definite) and the
# square root of the fiducial total matrices
Chat_tot = total_Cl_matrix(data, N_ell)
sqrt_fid = np.array([mat_pow(C_tot_true[il], 0.5) for il in range(n_ell_bins)])
iu_hl = np.triu_indices(n_freq)

def log_like_hl(pv):
    C_mod = total_Cl_matrix(
        compute_signal_model(**dict(zip(param_names, pv))), N_ell)
    x = np.empty((n_ell_bins, n_cross))
    for il in range(n_ell_bins):
        Cm_isqrt = mat_pow(C_mod[il], -0.5)
        w, V = np.linalg.eigh(Cm_isqrt @ Chat_tot[il] @ Cm_isqrt)
        if np.any(w <= 0):
            return -np.inf
        X = sqrt_fid[il] @ ((V * hl_g(w)) @ V.T) @ sqrt_fid[il]
        x[il] = X[iu_hl]
    xf = x.flatten()
    return -0.5 * xf @ inv_cov @ xf

def log_post_hl(pv):
    lp = log_prior(pv)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_like_hl(pv)

# Sanity check: near the truth, H&L should agree with the Gaussian chi^2
print(f'-2 lnL at truth:  Gaussian = {-2 * log_likelihood(p0_true):.1f},  '
      f'H&L = {-2 * log_like_hl(p0_true):.1f}')

print('\nH&L run:')
samples_hl = quick_mcmc(log_post_hl, n_steps_q=10000, n_burn_q=2000)

fig, ax = plt.subplots(figsize=(7, 4))
bins = np.linspace(0, 0.06, 40)
ax.hist(flat_samples[:, 0], bins=bins, density=True, alpha=0.6, color='C0',
        label='Gaussian likelihood')
ax.hist(samples_hl[:, 0], bins=bins, density=True, histtype='step', lw=2,
        color='C3', label='Hamimeche-Lewis likelihood')
ax.axvline(params_true['r'], color='r', ls='--', label='True $r$')
ax.set_xlabel('$r$'); ax.set_ylabel('Posterior density')
ax.legend()
plt.tight_layout()
plt.show()

print(f'sigma(r):  Gaussian = {sig68(flat_samples[:, 0]):.4f},  '
      f'H&L = {sig68(samples_hl[:, 0]):.4f}')
print('At Nmodes ~ 1000 per bin the two likelihoods are indistinguishable;')
print('H&L matters at low Nmodes (small f_sky, lowest multipoles).')
```
